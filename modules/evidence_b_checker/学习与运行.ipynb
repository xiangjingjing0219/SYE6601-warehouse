四态判定学习笔记

核验事项：本次故障是否达到60分钟门槛？

1. SUPPORT（支持）
有效记录显示90分钟，达到门槛。

2. REFUTE（反驳）
有效记录显示40分钟，没有达到门槛。

3. UNKNOWN（未知）
查询超时、403无权限、分页未查完，或没有适用的证据。
无法确认，不等于已经证明不满足。

4. CONFLICT（冲突）
有效证据分别显示90分钟和40分钟，
且按规则优先级仍无法解决。

判定顺序：
先检查证据是否适用，再根据有效证据判断四态。

注意：
达到时长门槛，不代表最终一定可以补偿。
还需要核查合同及免责条件。

````python
# 练习16：判断故障是否达到60分钟门槛
# 假设记录已经通过客户、版本和时间等有效性检查
incident_minutes = 60

if incident_minutes >= 60:
    print("SUPPORT：故障达到60分钟门槛")
else:
    print("REFUTE：故障没有达到60分钟门槛")
````

````python
# 练习17：没有时长记录时，输出 UNKNOWN
# None 表示这里暂时没有时长数据，不是0分钟
incident_minutes = None

if incident_minutes is None:
    print("UNKNOWN：没有故障时长记录，暂时无法判断")
elif incident_minutes >= 60:
    print("SUPPORT：故障达到60分钟门槛")
else:
    print("REFUTE：故障没有达到60分钟门槛")
````

故障时长检查小结

前提：记录已经通过适用性检查，门槛为60分钟。

时长 ≥ 60分钟 → SUPPORT
时长 < 60分钟 → REFUTE
没有时长数据（None）→ UNKNOWN

注意：None不是0。
None表示不知道时长；0表示明确记录为0分钟。
以上只判断时长门槛，不是最终补偿结论。

````python
# 练习18：用函数判断故障时长是否达到门槛
# 前提：证据已经通过适用性检查
# 判断时长门槛，并拒绝不合要求的数据
def check_duration(incident_minutes, threshold):
    # 门槛必须是大于0的数字，不能是True或False
    if type(threshold) not in (int, float):
        return "UNKNOWN"
    if not 0 < threshold < float("inf"):
        return "UNKNOWN"

    # 时长必须是非负数字，不能是文字、True或False
    if type(incident_minutes) not in (int, float):
        return "UNKNOWN"
    if not 0 <= incident_minutes < float("inf"):
        return "UNKNOWN"

    if incident_minutes >= threshold:
        return "SUPPORT"
    else:
        return "REFUTE"
````

元数据检查：合同版本

场景：
本次判断要求使用V2合同，但工具只返回了V1合同。

处理：
V1版本不适用于本次判断，不能作为支持或反驳的依据。
如果没有其他适用证据，相关待证明事项记为UNKNOWN。
原因码：STALE_VERSION（版本不适用）。

注意：
版本是否适用，要依据规则和相关时间确认，
不能仅凭版本号较小就认定无效。

````python
# 练习19：检查合同版本是否符合本次要求
# required_version 是本次判断要求使用的版本
def check_version(evidence_version, required_version):
    if evidence_version == required_version:
        return {
            "version_match": True,
            "reason": "版本符合要求，还需检查其他条件"
        }
    else:
        return {
            "version_match": False,
            "reason": "版本不适用，不能作为本次判断的依据"
        }

# 一次查看版本一致、不一致两种情况
print(check_version("V2", "V2"))
print(check_version("V1", "V2"))
````

````python
# 练习20：检查故障日期是否在合同有效期内
# 本练习约定：起止日期当天都有效
from datetime import date

def check_effective_time(incident_date, start_date, end_date):
    # 将 YYYY-MM-DD 格式的文字转换成日期
    incident = date.fromisoformat(incident_date)
    start = date.fromisoformat(start_date)
    end = date.fromisoformat(end_date)

    return start <= incident <= end

# 一次查看有效期内、尚未生效两种情况
print(check_effective_time("2026-06-01", "2026-01-01", "2026-12-31"))
print(check_effective_time("2025-12-01", "2026-01-01", "2026-12-31"))
````

时间字段的区别

effective_from / effective_to：证据的有效期。
fetched_at：系统取得证据的时间。

刚查到的合同，不代表合同当前有效。
应按业务规则要求的时间，检查合同是否适用。

````python
# 练习21：检查分页是否完成
def check_pagination(pagination_complete):
    if pagination_complete:
        return "分页已完成，还需结合返回内容和其他检查判断"
    else:
        return "分页未完成，不能仅凭空结果认定记录不存在"

# 一次查看两种情况
print(check_pagination(True))
print(check_pagination(False))
````

````python
# 练习22：检查工具调用状态
def check_tool_status(tool_status):
    if tool_status == "SUCCESS":
        return "调用成功，还需检查证据是否适用、完整"
    elif tool_status == "TIMEOUT":
        return "UNKNOWN：查询超时，未取得充分证据"
    elif tool_status == "ACCESS_DENIED":
        return "UNKNOWN：无权限查看，记录原因供Loop组转人工"
    else:
        return "UNKNOWN：无法识别调用状态，需要核实"

# 一次查看三种情况
print(check_tool_status("SUCCESS"))
print(check_tool_status("TIMEOUT"))
print(check_tool_status("ACCESS_DENIED"))
````

元数据检查阶段小结

已练习：客户编号、合同版本、有效期、取得时间、
分页完整性和工具调用状态。

检查原则：
1. 先确认材料是否适用，再判断支持或反驳。
2. 查询成功，不代表证据充分。
3. 超时、无权限或分页未完，不能当作“记录不存在”。

````python
# 练习23：检查是否提供了客户编号
def check_customer_field(evidence):
    customer_id = evidence.get("customer_id")

    if customer_id is None:
        return "UNKNOWN：缺少客户编号，需要核实"
    else:
        return "已提供客户编号，还需检查是否与当前客户一致"

# 一次查看字段存在、字段缺失两种情况
print(check_customer_field({"customer_id": "C001"}))
print(check_customer_field({}))
````

````python
# 练习24：检查证据的服务范围
# 入门练习：每条证据只有一个服务范围，要求完全一致
def check_scope(evidence_scope, required_scope):
    if evidence_scope is None:
        return "UNKNOWN：缺少适用范围，需要核实"
    elif evidence_scope == required_scope:
        return "范围一致，还需检查其他条件"
    else:
        return "范围不适用，不能作为本次判断的依据"

# 一次查看范围一致、不一致和缺失三种情况
print(check_scope("CLOUD_SERVER", "CLOUD_SERVER"))
print(check_scope("EMAIL", "CLOUD_SERVER"))
print(check_scope(None, "CLOUD_SERVER"))
````

结构化模型学习笔记

作用：规定证据必须有哪些字段，每个字段应填写什么类型。

例如：
customer_id：客户编号，应为字符串（文字）。
pagination_complete：分页是否完成，应为布尔值（True或False）。

结构检查通过，只代表填写符合要求；
还要继续检查客户、版本、时间等是否适用于本次任务。

````python
# 练习25：规定证据字段及其类型
from pydantic import BaseModel, StrictStr, StrictBool

class Evidence(BaseModel):
    # 客户编号必须是文字
    customer_id: StrictStr

    # 分页是否完成必须是 True 或 False
    pagination_complete: StrictBool

# 按要求填写一张证据资料卡
evidence = Evidence(
    customer_id="C001",
    pagination_complete=False
)

print(evidence)
````

````python
# 练习26：缺少必需字段时，显示清楚的提示
from pydantic import ValidationError

try:
    # 故意不填写 pagination_complete
    evidence = Evidence(customer_id="C001")
    print("格式检查通过")
except ValidationError:
    print("格式检查未通过：缺少必需的分页状态字段")
````

````python
# 练习27：分页状态填成文字时，显示校验提示
try:
    evidence = Evidence(
        customer_id="C001",
        pagination_complete="C001"
    )
    print("格式检查通过")
except ValidationError:
    print("格式检查未通过：分页状态必须填写 True 或 False")
````

案例1：有效专属合同优先

待核验事项：本次故障是否符合补偿条件？
通用政策：支持补偿。
专属合同：明确排除本次补偿，且适用性检查通过。
优先级：专属合同优先。

预期状态：REFUTE。
原因：按优先级采用有效专属合同，反驳补偿资格。

案例2：专属合同在故障发生时已过期

待核验事项：本次故障是否符合补偿条件？
通用政策：支持补偿。
已取得的专属合同：已过期，不能用于本次判断。
适用合同是否存在及其内容：尚未查清。

预期状态：UNKNOWN。
原因：缺少适用的关键合同证据。
待补证：故障发生时适用的客户专属合同，
或完整、有效的查询结果，确认不存在适用专属合同。

案例3：无权限取得专属合同

待核验事项：本次故障是否符合补偿条件？
通用政策：支持补偿。
专属合同查询：返回403，没有取得合同。

预期状态：UNKNOWN。
原因码：ACCESS_DENIED。
原因：无法核实关键合同，不能肯定补偿。
交接：将缺口和无权限原因交给Loop组，按约定转人工。

````python
# 练习30：整合客户编号检查，返回结果和原因码
def check_customer_metadata(evidence, current_customer):
    customer_id = evidence.get("customer_id")

    if customer_id is None:
        return {
            "customer_match": None,
            "reason_code": "MISSING_CUSTOMER_ID"
        }
    elif customer_id != current_customer:
        return {
            "customer_match": False,
            "reason_code": "WRONG_CUSTOMER"
        }
    else:
        return {
            "customer_match": True,
            "reason_code": "CUSTOMER_MATCH"
        }

# 一次查看：一致、不一致、缺失
print(check_customer_metadata({"customer_id": "C001"}, "C001"))
print(check_customer_metadata({"customer_id": "C002"}, "C001"))
print(check_customer_metadata({}, "C001"))
````

````python
# 练习31：检查工具状态，返回结果和原因码
def check_tool_metadata(evidence):
    tool_status = evidence.get("tool_status")

    if tool_status == "SUCCESS":
        return {
            "tool_success": True,
            "reason_code": "TOOL_SUCCESS"
        }
    elif tool_status == "TIMEOUT":
        return {
            "tool_success": False,
            "reason_code": "TIMEOUT"
        }
    elif tool_status == "ACCESS_DENIED":
        return {
            "tool_success": False,
            "reason_code": "ACCESS_DENIED"
        }
    else:
        return {
            "tool_success": None,
            "reason_code": "MISSING_OR_UNKNOWN_TOOL_STATUS"
        }

# 一次查看成功、超时、无权限三种情况
print(check_tool_metadata({"tool_status": "SUCCESS"}))
print(check_tool_metadata({"tool_status": "TIMEOUT"}))
print(check_tool_metadata({"tool_status": "ACCESS_DENIED"}))
````

````python
# 练习33：检查分页字段，返回结果和原因码
def check_pagination_metadata(evidence):
    complete = evidence.get("pagination_complete")

    if complete is True:
        return {
            "pagination_complete": True,
            "reason_code": "PAGINATION_COMPLETE"
        }
    elif complete is False:
        return {
            "pagination_complete": False,
            "reason_code": "PARTIAL_PAGE"
        }
    else:
        return {
            "pagination_complete": None,
            "reason_code": "MISSING_OR_INVALID_PAGINATION"
        }

# 一次查看：查完、未查完、缺少字段
print(check_pagination_metadata({"pagination_complete": True}))
print(check_pagination_metadata({"pagination_complete": False}))
print(check_pagination_metadata({}))
````

````python
# 练习32（更新）：汇总客户、工具状态和分页检查
def check_metadata_basic(evidence, current_customer):
    return {
        "customer_check": check_customer_metadata(
            evidence, current_customer
        ),
        "tool_check": check_tool_metadata(evidence),
        "pagination_check": check_pagination_metadata(evidence)
    }

# 模拟：客户一致、调用成功，但分页未完成
sample_evidence = {
    "customer_id": "C001",
    "tool_status": "SUCCESS",
    "pagination_complete": False
}

result = check_metadata_basic(sample_evidence, "C001")
print(json.dumps(result, ensure_ascii=False, indent=2))
````

````python
# 练习34：检查版本，返回结果和原因码
def check_version_metadata(evidence, required_version):
    version = evidence.get("version")

    if version is None:
        return {
            "version_match": None,
            "reason_code": "MISSING_VERSION"
        }
    elif version != required_version:
        return {
            "version_match": False,
            "reason_code": "STALE_VERSION"
        }
    else:
        return {
            "version_match": True,
            "reason_code": "VERSION_MATCH"
        }

# 本次要求V2，一次查看一致、不一致和缺失
print(check_version_metadata({"version": "V2"}, "V2"))
print(check_version_metadata({"version": "V1"}, "V2"))
print(check_version_metadata({}, "V2"))
````

````python
# 练习35：检查服务范围，返回结果和原因码
# 练习约定：范围为单个字符串，要求完全一致
def check_scope_metadata(evidence, required_scope):
    scope = evidence.get("scope")

    if scope is None:
        return {
            "scope_match": None,
            "reason_code": "MISSING_SCOPE"
        }
    elif scope != required_scope:
        return {
            "scope_match": False,
            "reason_code": "WRONG_SCOPE"
        }
    else:
        return {
            "scope_match": True,
            "reason_code": "SCOPE_MATCH"
        }

# 当前核查云服务器服务
print(check_scope_metadata({"scope": "CLOUD_SERVER"}, "CLOUD_SERVER"))
print(check_scope_metadata({"scope": "EMAIL"}, "CLOUD_SERVER"))
print(check_scope_metadata({}, "CLOUD_SERVER"))
````

````python
# 练习36：汇总五项元数据检查
def check_metadata(evidence, current_customer,
                   required_version, required_scope):
    return {
        "customer_check": check_customer_metadata(
            evidence, current_customer
        ),
        "tool_check": check_tool_metadata(evidence),
        "pagination_check": check_pagination_metadata(evidence),
        "version_check": check_version_metadata(
            evidence, required_version
        ),
        "scope_check": check_scope_metadata(
            evidence, required_scope
        )
    }

# 模拟：其他项目符合要求，但版本不适用
sample_evidence = {
    "customer_id": "C001",
    "tool_status": "SUCCESS",
    "pagination_complete": True,
    "version": "V1",
    "scope": "CLOUD_SERVER"
}

result = check_metadata(
    sample_evidence, "C001", "V2", "CLOUD_SERVER"
)

print(json.dumps(result, ensure_ascii=False, indent=2))
````

````python
# 练习37：检查有效期，返回结果和原因码
from datetime import date

def check_time_metadata(evidence, incident_date):
    start = evidence.get("effective_from")
    end = evidence.get("effective_to")

    if start is None or end is None:
        return {
            "time_valid": None,
            "reason_code": "MISSING_EFFECTIVE_TIME"
        }

    try:
        incident = date.fromisoformat(incident_date)
        start = date.fromisoformat(start)
        end = date.fromisoformat(end)
    except (TypeError, ValueError):
        return {
            "time_valid": None,
            "reason_code": "INVALID_DATE_FORMAT"
        }

    # 起止日期写反时，不能继续判断
    if start > end:
        return {
            "time_valid": None,
            "reason_code": "INVALID_TIME_RANGE"
        }

    if start <= incident <= end:
        return {
            "time_valid": True,
            "reason_code": "TIME_VALID"
        }

    return {
        "time_valid": False,
        "reason_code": "OUTSIDE_EFFECTIVE_TIME"
    }

# 本练习使用YYYY-MM-DD日期，包含起止当天
sample_contract = {
    "effective_from": "2026-01-01",
    "effective_to": "2026-12-31"
}

print(check_time_metadata(sample_contract, "2026-06-01"))
print(check_time_metadata(sample_contract, "2025-12-01"))
````

````python
# 练习38：在五项检查结果中加入有效期检查
sample_evidence = {
    "customer_id": "C001",
    "tool_status": "SUCCESS",
    "pagination_complete": True,
    "version": "V2",
    "scope": "CLOUD_SERVER",
    "effective_from": "2026-01-01",
    "effective_to": "2026-12-31"
}

# 先执行已有的五项检查
result = check_metadata(
    sample_evidence, "C001", "V2", "CLOUD_SERVER"
)

# 再加入时间检查：故障发生在合同生效之前
result["time_check"] = check_time_metadata(
    sample_evidence, "2025-12-01"
)

print(json.dumps(result, ensure_ascii=False, indent=2))
````

````python
# 练习39：收集六项检查中未通过或无法确认的原因
def collect_metadata_issues(checks):
    fields = {
        "customer_check": "customer_match",
        "tool_check": "tool_success",
        "pagination_check": "pagination_complete",
        "version_check": "version_match",
        "scope_check": "scope_match",
        "time_check": "time_valid"
    }

    issues = []

    # 逐项查看检查结果
    for check_name, result_field in fields.items():
        item = checks.get(check_name, {})

        # 只有明确为 True 才通过；False和None都记录
        if item.get(result_field) is not True:
            issues.append(
                item.get("reason_code", "MISSING_CHECK_RESULT")
            )

    return issues

# 使用上一格的六项检查结果
print(collect_metadata_issues(result))
````

````python
# 练习40：汇总六项元数据检查是否全部通过
def summarize_metadata(checks):
    issues = collect_metadata_issues(checks)

    return {
        "metadata_passed": len(issues) == 0,
        "reason_codes": issues
    }

print(json.dumps(
    summarize_metadata(result),
    ensure_ascii=False,
    indent=2
))
````

````python
# 练习41：先检查元数据，再校验时长并说明原因
def check_duration_with_metadata(checks, incident_minutes, threshold):
    summary = summarize_metadata(checks)

    if not summary["metadata_passed"]:
        return {
            "status": "UNKNOWN",
            "reason_codes": summary["reason_codes"]
        }

    # 检查门槛
    if type(threshold) not in (int, float):
        reason = "INVALID_THRESHOLD"
    elif not 0 < threshold < float("inf"):
        reason = "INVALID_THRESHOLD"

    # 区分时长缺失与填写错误
    elif incident_minutes is None:
        reason = "MISSING_INCIDENT_DURATION"
    elif type(incident_minutes) not in (int, float):
        reason = "INVALID_INCIDENT_DURATION"
    elif not 0 <= incident_minutes < float("inf"):
        reason = "INVALID_INCIDENT_DURATION"
    else:
        status = check_duration(incident_minutes, threshold)
        return {
            "status": status,
            "reason_codes": [
                "DURATION_THRESHOLD_MET"
                if status == "SUPPORT"
                else "DURATION_BELOW_THRESHOLD"
            ]
        }

    return {
        "status": "UNKNOWN",
        "reason_codes": [reason]
    }
````

````python
# 练习42：检查同一条故障记录，再判断时长
incident_evidence = {
    "customer_id": "C001",
    "tool_status": "SUCCESS",
    "pagination_complete": True,
    "version": "V2",
    "scope": "CLOUD_SERVER",
    "effective_from": "2026-01-01",
    "effective_to": "2026-12-31",
    "incident_minutes": 40
}

checks = check_metadata(
    incident_evidence, "C001", "V2", "CLOUD_SERVER"
)
checks["time_check"] = check_time_metadata(
    incident_evidence, "2026-06-01"
)

decision = check_duration_with_metadata(
    checks, incident_evidence["incident_minutes"], 60
)

print(json.dumps(decision, ensure_ascii=False, indent=2))
````

````python
# 练习45：汇总同一待证明事项的有效证据

# 使用前提：
# 1. 判断的是同一个待证明事项
# 2. 证据已经通过适用性检查，相关查询已完整
# 3. 已按业务规则处理优先级
# 4. 仍存在无法消解的有效相反证据，才判为CONFLICT
# 当前函数接收手动填写的布尔值，尚未自动处理原始证据

def combine_evidence(has_support, has_refute):
    if has_support and has_refute:
        return "CONFLICT"
    elif has_support:
        return "SUPPORT"
    elif has_refute:
        return "REFUTE"
    else:
        return "UNKNOWN"

# 一次查看四种组合
print(combine_evidence(True, False))
print(combine_evidence(False, True))
print(combine_evidence(False, False))
print(combine_evidence(True, True))
````

# 第二阶段：把练习接成一个小 Checker

我们先做一个可运行的小闭环：**输入任务和模拟记录 → 检查材料 → 判断时长 → 输出缺口和引用 → 补入记录再检查**。

这次不自动决定维护扣除、合同优先级、历史版本选择。由调用者明确给出本次需要的版本和核验时间；这些演示约定仍需 A 确认。

## 第 1 部分：准备工具和约定
只用 Python 自带功能。新函数使用不同名字，不覆盖上面的练习函数。日期要求带时区，避免把北京时间误当成 UTC。

````python
# 导入工具：JSON负责数据展示；deepcopy用于复制样例，避免改坏原数据。
import json
from copy import deepcopy
from datetime import datetime
import math

def parse_checker_time(value):
    # 要求明确的时区，例如 2026-06-01T00:00:00Z。
    if not isinstance(value, str):
        raise ValueError("时间必须是带时区的文字")
    dt = datetime.fromisoformat(value.replace("Z", "+00:00"))
    if dt.tzinfo is None:
        raise ValueError("时间缺少时区，请补充 Z 或 +08:00")
    return dt

def valid_number(value, positive=False):
    # bool 在 Python 中属于数字的子类，因此必须主动排除。
    return (type(value) in (int, float) and math.isfinite(value)
            and (value > 0 if positive else value >= 0))
````

## 第 2 部分：检查一条材料是否能用

新增检查来源、故障编号和引用，防止把同一客户另一场故障拿来判断。有效期按调用者传入的核验时间检查，演示约定包含两端。

工具成功与分页完整是两件事。这个保守原型要求完整分页；是否允许凭一条决定性记录提前判断，后续由 A 确认。

返回的 `checks` 对应置信度组的五项检查。完整原始记录另行保留，不会只剩一个分数。

````python
def inspect_duration_record(record, task, config):
    # 每项从“尚未确认”开始，不能默认合格。
    checks = dict.fromkeys(["source_allowed", "scope_valid", "version_valid",
                            "time_valid", "query_sufficient"], None)
    reasons = []
    if not isinstance(record, dict):
        return checks, ["INVALID_RECORD"]

    checks["source_allowed"] = record.get("source_type") == config["source_type"]
    checks["scope_valid"] = all(record.get(k) == task[k]
                                  for k in ["customer_id", "scope", "incident_id"])
    checks["version_valid"] = record.get("version") == config["required_version"]
    checks["query_sufficient"] = (record.get("tool_status") == "SUCCESS"
                                  and record.get("pagination_complete") is True)
    try:
        start = parse_checker_time(record.get("effective_from"))
        end = parse_checker_time(record.get("effective_to"))
        at = parse_checker_time(task["reference_time"])
        checks["time_valid"] = start <= at <= end
        if start > end:
            reasons.append("INVALID_TIME_RANGE")
    except (ValueError, TypeError):
        checks["time_valid"] = None
        reasons.append("INVALID_OR_MISSING_TIME")

    # 抓取时间仅校验格式；尚未确认资料多久过期，不自创天数限制。
    try:
        parse_checker_time(record.get("fetched_at"))
    except (ValueError, TypeError):
        reasons.append("INVALID_FETCHED_AT")
    reason_map = {"source_allowed": "WRONG_SOURCE", "scope_valid": "WRONG_SCOPE_OR_INCIDENT",
                  "version_valid": "WRONG_VERSION", "time_valid": "OUTSIDE_EFFECTIVE_TIME",
                  "query_sufficient": "INCOMPLETE_QUERY"}
    reasons += [reason_map[k] for k, value in checks.items() if value is not True]
    if record.get("tool_status") in ("TIMEOUT", "ACCESS_DENIED"):
        reasons.append(record["tool_status"])
    for key in ["evidence_id", "citation"]:
        if not isinstance(record.get(key), str) or not record[key].strip():
            reasons.append("MISSING_" + key.upper())
    return checks, sorted(set(reasons))
````

## 第 3 部分：自动整理多条证据、原因和缺口

这里合并的是**同一次故障的同一个时长命题**。不处理不同合同优先级。

- 有合格材料且时长达标：SUPPORT。
- 有合格材料且时长不足：REFUTE。
- 没有可用材料：UNKNOWN。
- 合格材料得出相反结果：CONFLICT。

一条坏材料不会抹掉另一条充分有效材料；坏材料保留在审查记录中。完整补偿还缺其他检查，因此 `current_decision` 始终为 `None`。
输入结构错误会报可读错误，不伪装成正常业务结论。

````python
def run_duration_checker(task, records, config):
    # 先检查调用者填写的任务和配置。
    for obj, keys in [(task, ["task_id", "customer_id", "scope", "incident_id", "reference_time"]),
                      (config, ["obligation_id", "rule_id", "source_type", "required_version", "rules_version"])]:
        if not isinstance(obj, dict):
            raise ValueError("任务和配置必须是字典")
        for key in keys:
            if not isinstance(obj.get(key), str) or not obj[key].strip():
                raise ValueError("缺少或填错字段：" + key)
    parse_checker_time(task["reference_time"])
    if not valid_number(config.get("threshold"), positive=True):
        raise ValueError("threshold 必须是大于0的有限数字")
    if not isinstance(records, list):
        raise ValueError("records 必须是列表")
    # 同一编号不能代表两份不同材料，避免引用混乱。
    ids = [r.get("evidence_id") for r in records if isinstance(r, dict) and r.get("evidence_id")]
    if any(not isinstance(i, str) for i in ids) or len(ids) != len(set(ids)):
        raise ValueError("证据编号必须是文字且不能重复")

    audits, accepted, states = [], [], set()
    for record in records:
        checks, reasons = inspect_duration_record(record, task, config)
        minutes = record.get("incident_minutes") if isinstance(record, dict) else None
        if minutes is None:
            reasons.append("MISSING_INCIDENT_DURATION")
        elif not valid_number(minutes):
            reasons.append("INVALID_INCIDENT_DURATION")
        usable = not reasons
        audits.append({"record": deepcopy(record), "checks": checks,
                       "usable": usable, "reason_codes": sorted(set(reasons))})
        if usable:
            accepted.append(record)
            states.add("SUPPORT" if minutes >= config["threshold"] else "REFUTE")

    status = "CONFLICT" if len(states) == 2 else next(iter(states), "UNKNOWN")
    reason_map = {"SUPPORT": "DURATION_THRESHOLD_MET", "REFUTE": "DURATION_BELOW_THRESHOLD",
                  "CONFLICT": "CONTRADICTORY_VALID_RECORDS", "UNKNOWN": "NO_USABLE_DURATION_EVIDENCE"}
    reason_codes = [reason_map[status]]
    if status == "UNKNOWN":
        reason_codes += sorted({r for a in audits for r in a["reason_codes"]})
    reason = {"SUPPORT": "有效材料显示时长达到演示门槛。", "REFUTE": "有效材料显示时长不足演示门槛。",
              "CONFLICT": "有效材料对同一时长门槛给出相反结果，需核实。",
              "UNKNOWN": "尚无充分有效的故障时长材料。"}[status]
    obligation = {"obligation_id": config["obligation_id"], "status": status,
                  "reason_codes": reason_codes, "reason": reason,
                  "evidence_ids": [r["evidence_id"] for r in accepted]}
    gaps = []
    if status in ("UNKNOWN", "CONFLICT"):
        gaps.append({"obligation_id": config["obligation_id"], "source_type": config["source_type"],
                     "customer_id": task["customer_id"], "incident_id": task["incident_id"],
                     "reason_codes": reason_codes, "required_for_decision": True})
    # 无论时长结果如何，都明确说明整个业务模型尚不完整。
    gaps.append({"obligation_id": "DEMO-PENDING-BUSINESS", "source_type": None,
                 "reason_codes": ["BUSINESS_MODEL_PENDING"], "required_for_decision": True})
    return {"task_id": task["task_id"], "synthetic_demo": True,
            "obligation_results": [obligation], "gaps": gaps,
            "current_decision": None, "audits": audits,
            "evidence": deepcopy(records)}
````

## 第 4 部分：先跑一份正常样例
下面所有编号、门槛和记录都是演示值，不冒充 A 组正式业务规则。可以把时长 90 改为 40，看第一项怎样变化。

````python
demo_task = {"task_id": "DEMO-T001", "customer_id": "C001", "incident_id": "INC-001",
             "scope": "CLOUD_SERVER", "reference_time": "2026-06-01T00:00:00Z"}
demo_config = {"obligation_id": "DEMO-O-DURATION", "rule_id": "DEMO-RULE",
               "source_type": "MONITORING_SYSTEM", "required_version": "V2",
               "rules_version": "demo-only-not-01a", "threshold": 60}
demo_record = {"evidence_id": "DEMO-E001", "customer_id": "C001", "incident_id": "INC-001",
               "scope": "CLOUD_SERVER", "version": "V2", "source_type": "MONITORING_SYSTEM",
               "effective_from": "2026-01-01T00:00:00Z", "effective_to": "2026-12-31T23:59:59Z",
               "fetched_at": "2026-06-02T00:00:00Z", "tool_status": "SUCCESS",
               "pagination_complete": True, "citation": "demo://INC-001/E001", "incident_minutes": 90}
example_result = run_duration_checker(demo_task, [demo_record], demo_config)
print(json.dumps(example_result["obligation_results"], ensure_ascii=False, indent=2))
print("完整补偿结论：", example_result["current_decision"])
````

## 第 5 部分：一次运行边界测试
这些是开发自测，不是测试组的独立标准答案，也不是模型准确率。只需一次 CPU 运行，不消耗 GPU 配额。

````python
def changed_record(**updates):
    # 复制后修改，原始正常样例保持可复用。
    result = deepcopy(demo_record)
    result.update(updates)
    return result

cases = [
    ("正常达标", [demo_record], "SUPPORT"),
    ("不足门槛", [changed_record(incident_minutes=40)], "REFUTE"),
    ("刚好门槛", [changed_record(incident_minutes=60)], "SUPPORT"),
    ("没有记录", [], "UNKNOWN"),
    ("时长缺失", [changed_record(incident_minutes=None)], "UNKNOWN"),
    ("负数", [changed_record(incident_minutes=-1)], "UNKNOWN"),
    ("布尔值", [changed_record(incident_minutes=True)], "UNKNOWN"),
    ("文字数字", [changed_record(incident_minutes="40")], "UNKNOWN"),
    ("无穷大", [changed_record(incident_minutes=float("inf"))], "UNKNOWN"),
    ("客户错误", [changed_record(customer_id="C002")], "UNKNOWN"),
    ("另一场故障", [changed_record(incident_id="INC-002")], "UNKNOWN"),
    ("版本不符", [changed_record(version="V1")], "UNKNOWN"),
    ("合同期外", [changed_record(effective_to="2026-02-01T00:00:00Z")], "UNKNOWN"),
    ("无时区", [changed_record(effective_from="2026-01-01")], "UNKNOWN"),
    ("未查完分页", [changed_record(pagination_complete=False)], "UNKNOWN"),
    ("超时", [changed_record(tool_status="TIMEOUT")], "UNKNOWN"),
    ("无权限", [changed_record(tool_status="ACCESS_DENIED")], "UNKNOWN"),
    ("无引用", [changed_record(citation="")], "UNKNOWN"),
    ("来源错误", [changed_record(source_type="OTHER")], "UNKNOWN"),
    ("相反有效证据", [demo_record, changed_record(evidence_id="DEMO-E002", incident_minutes=40)], "CONFLICT"),
    ("坏材料不抹掉好材料", [demo_record, changed_record(evidence_id="DEMO-E002", customer_id="C002")], "SUPPORT"),
]
for name, records, expected in cases:
    result = run_duration_checker(demo_task, records, demo_config)
    actual = result["obligation_results"][0]["status"]
    assert actual == expected, (name, expected, actual)
    assert result["current_decision"] is None
print(f"{len(cases)} 个业务边界自测全部通过。")
# 再验证不修改输入，以及同输入同输出。
before = deepcopy(demo_record)
assert run_duration_checker(demo_task, [demo_record], demo_config) == run_duration_checker(demo_task, [demo_record], demo_config)
assert before == demo_record
try:
    run_duration_checker(demo_task, [demo_record, demo_record], demo_config)
except ValueError as error:
    print("重复编号按预期被拒绝：", error)
else:
    raise AssertionError("重复编号没有被拒绝")
````

## 第 6 部分：补查前后重新检查
这只是模拟 Loop 把新材料交回来，不是已经实现自动查询、重试和人工升级。第一次没查到时长，第二次交入正常记录，观察时长状态变化。

````python
before_result = run_duration_checker(demo_task, [], demo_config)
after_result = run_duration_checker(demo_task, [demo_record], demo_config)
print("补材料前：", before_result["obligation_results"][0]["status"])
print("补材料后：", after_result["obligation_results"][0]["status"])
print("补材料后仍有的缺口：", after_result["gaps"])
````

## 第 7 部分：转换为置信度 A 的接口草案
团队评分器接收五项 checks 和证据索引。我们只把已经查明的时长转换过去；另放一个明显的“业务模型待确认”占位义务，防止只检查时长就显示证据完整。

这不是正式义务清单。它是显式的演示投影，不能用于报告正式完整率。流程状态先用 RUNNING；真正的停止或升级由 Loop 提供。

````python
def to_confidence_draft(result, config, snapshot_id):
    item = result["obligation_results"][0]
    # 只对已使用的有效记录声明五项通过，不把失败记录说成有效。
    values = True if item["evidence_ids"] else None
    checks = dict.fromkeys(["source_allowed", "scope_valid", "version_valid",
                            "time_valid", "query_sufficient"], values)
    obligation = {**deepcopy(item), "rule_id": config["rule_id"],
                  "statement": "本次故障时长达到演示门槛", "applicability": "APPLICABLE",
                  "applicability_reason": "仅为时长原型演示，非正式业务适用性结论。",
                  "required_for_decision": True, "must_verify": False, "checks": checks}
    pending = {"obligation_id": "DEMO-PENDING-BUSINESS", "rule_id": "DEMO-RULE",
               "statement": "完整业务义务清单及规则口径已经确认", "applicability": "UNDETERMINED",
               "applicability_reason": "等待A组确认正式业务模型；这里只是占位，不能充当正式分母。",
               "required_for_decision": True, "must_verify": True, "status": "UNKNOWN",
               "checks": dict.fromkeys(checks, None), "evidence_ids": [],
               "reason_codes": ["BUSINESS_MODEL_PENDING"], "reason": "维护、合同优先关系和完整义务待确认。"}
    evidence_index = [{k: a["record"][k] for k in ["evidence_id", "source_type", "citation"]}
                      for a in result["audits"] if a["usable"]]
    return {"schema_version": "0.1-draft", "task_id": result["task_id"],
            "run_id": "DEMO-RUN-001", "snapshot_id": snapshot_id,
            "rules_version": config["rules_version"], "synthetic_demo": True,
            "checker_decision": "UNDETERMINED", "process_state": "RUNNING", "stop_reason": None,
            "obligations": [obligation, pending], "evidence": evidence_index}

score_snapshot_draft = to_confidence_draft(after_result, demo_config, "DEMO-SNAPSHOT-AFTER")
print(json.dumps(score_snapshot_draft, ensure_ascii=False, indent=2))
````

````python
import json
import hashlib
from pathlib import Path
from copy import deepcopy

# 下面两段是A组文件的原样数据副本，不是自行编写的新业务规则。
A_RULES_COPY = json.loads(r'''[
  {
    "rule_id": "RULE-001",
    "rule_name": "VIP客户专属SLA补偿规则",
    "description": "针对购买企业VIP套餐的客户，故障满30分钟即可申请赔偿，但需扣除计划内维护；若因客户配置错误或不可抗力导致的故障属于免责例外。",
    "priority": 1,
    "effective_time": {
      "start": "2026-01-01T00:00:00Z",
      "end": "2026-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": [
        "VIP"
      ],
      "plan_type": [
        "Enterprise_Pro"
      ]
    },
    "thresholds": {
      "min_incident_minutes": 30
    },
    "exceptions": [
      "CLIENT_CONFIG_ERROR",
      "FORCE_MAJEURE"
    ]
  },
  {
    "rule_id": "RULE-002",
    "rule_name": "通用客户标准补偿政策",
    "description": "针对普通基础套餐客户，无专属合同下降级适用；故障满60分钟可赔偿；同样排除客户自身原因引发的故障。",
    "priority": 100,
    "effective_time": {
      "start": "2025-01-01T00:00:00Z",
      "end": "2027-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": [
        "STANDARD",
        "BASIC"
      ],
      "plan_type": [
        "ALL"
      ]
    },
    "thresholds": {
      "min_incident_minutes": 60
    },
    "exceptions": [
      "CLIENT_CONFIG_ERROR"
    ]
  },
  {
    "rule_id": "RULE-003",
    "rule_name": "金融专区高可用保障规则",
    "description": "针对金融专区套餐客户，故障达到15分钟即可赔偿；排除黑客攻击与客户自身配置错误引发的故障。",
    "priority": 5,
    "effective_time": {
      "start": "2026-01-01T00:00:00Z",
      "end": "2026-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": [
        "FINANCE"
      ],
      "plan_type": [
        "FINANCE_ZONE"
      ]
    },
    "thresholds": {
      "min_incident_minutes": 15
    },
    "exceptions": [
      "DDoS_ATTACK",
      "CLIENT_CONFIG_ERROR"
    ]
  },
  {
    "rule_id": "RULE-004",
    "rule_name": "战略合作伙伴大客户极速补偿规则",
    "description": "针对战略合作大客户（KA），故障达10分钟即触发补偿保障，优先级别极高；仅排除客户人为主动破坏情况。",
    "priority": 2,
    "effective_time": {
      "start": "2026-01-01T00:00:00Z",
      "end": "2028-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": [
        "KA_STRATEGIC"
      ],
      "plan_type": [
        "Custom_Dedicated"
      ]
    },
    "thresholds": {
      "min_incident_minutes": 10
    },
    "exceptions": [
      "CLIENT_MALICIOUS_ACTION"
    ]
  },
  {
    "rule_id": "RULE-005",
    "rule_name": "免费试用与沙盒环境免赔规则",
    "description": "针对使用免费试用版或沙盒测试环境的客户，不提供SLA赔偿承诺，故障门槛设为不适用（极高值），所有故障均属于免赔范围。",
    "priority": 10,
    "effective_time": {
      "start": "2025-01-01T00:00:00Z",
      "end": "2099-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": [
        "TRIAL",
        "SANDBOX"
      ],
      "plan_type": [
        "Free_Tier"
      ]
    },
    "thresholds": {
      "min_incident_minutes": 999999
    },
    "exceptions": [
      "ALL_EXCEPTIONS"
    ]
  }
]''')
A_OBLIGATIONS_COPY = json.loads(r'''[
  {
    "rule_id": "RULE-001",
    "proof_obligations": [
      {
        "required_sources": [
          "CRM_CONTRACT_DB"
        ],
        "dependencies": {
          "prerequisite_step": null,
          "on_not_found": "PROCEED_TO_GENERAL_POLICY",
          "on_error_or_unchecked": "BLOCK_AND_RETURN_NULL"
        }
      },
      {
        "required_sources": [
          "MONITORING_SYSTEM",
          "MAINTENANCE_CALENDAR"
        ],
        "dependencies": {
          "prerequisite_step": "CHECK_EXCLUSIVE_CONTRACT"
        }
      }
    ]
  },
  {
    "rule_id": "RULE-002",
    "proof_obligations": [
      {
        "required_sources": [
          "CRM_CONTRACT_DB"
        ],
        "dependencies": {
          "prerequisite_step": null,
          "on_not_found": "APPLY_GENERAL_POLICY",
          "on_error_or_unchecked": "BLOCK_AND_RETURN_NULL"
        }
      },
      {
        "required_sources": [
          "MONITORING_SYSTEM"
        ],
        "dependencies": {
          "prerequisite_step": "VERIFY_NO_EXCLUSIVE_CONTRACT"
        }
      }
    ]
  },
  {
    "rule_id": "RULE-003",
    "proof_obligations": [
      {
        "required_sources": [
          "FINANCE_ZONE_DB"
        ],
        "dependencies": {
          "prerequisite_step": null,
          "on_not_found": "BLOCK_AND_RETURN_NULL",
          "on_error_or_unchecked": "BLOCK_AND_RETURN_NULL"
        }
      },
      {
        "required_sources": [
          "MONITORING_SYSTEM"
        ],
        "dependencies": {
          "prerequisite_step": "VERIFY_FINANCE_ZONE_IDENTITY"
        }
      }
    ]
  },
  {
    "rule_id": "RULE-004",
    "proof_obligations": [
      {
        "required_sources": [
          "KA_CLIENT_DB"
        ],
        "dependencies": {
          "prerequisite_step": null,
          "on_not_found": "PROCEED_TO_VIP_OR_GENERAL",
          "on_error_or_unchecked": "BLOCK_AND_RETURN_NULL"
        }
      },
      {
        "required_sources": [
          "MONITORING_SYSTEM",
          "TICKET_SYSTEM"
        ],
        "dependencies": {
          "prerequisite_step": "CHECK_KA_STATUS"
        }
      }
    ]
  },
  {
    "rule_id": "RULE-005",
    "proof_obligations": [
      {
        "required_sources": [
          "TRIAL_ACCOUNTS_DB"
        ],
        "dependencies": {
          "prerequisite_step": null,
          "on_not_found": "PROCEED_TO_STANDARD_RULES",
          "on_error_or_unchecked": "BLOCK_AND_RETURN_NULL"
        }
      },
      {
        "required_sources": [
          "SANDBOX_LOGS"
        ],
        "dependencies": {
          "prerequisite_step": "CHECK_TRIAL_STATUS"
        }
      }
    ]
  }
]''')

def load_a_files(folder="."):
    paths = [Path(folder) / "rules.json", Path(folder) / "obligations.json"]
    if any(p.exists() for p in paths) and not all(p.exists() for p in paths):
        raise ValueError("请同时提供 rules.json 和 obligations.json，避免混用版本")
    if all(p.exists() for p in paths):
        try:
            data = [json.loads(p.read_text(encoding="utf-8-sig")) for p in paths]
        except (json.JSONDecodeError, UnicodeError) as e:
            raise ValueError("A组文件不是可读取的UTF-8 JSON：" + str(e)) from e
        origin = "当前文件夹上传的A组文件"
    else:
        data = [deepcopy(A_RULES_COPY), deepcopy(A_OBLIGATIONS_COPY)]
        origin = "此前提供的A组文件副本；请确认是否有会后更新"
    # 指纹用于区分文件内容，不冒充A组批准的版本号。
    digest = hashlib.sha256(json.dumps(data, ensure_ascii=False, sort_keys=True).encode()).hexdigest()[:16]
    return {"rules": data[0], "obligations": data[1], "origin": origin,
            "content_fingerprint": digest}

a_bundle = load_a_files()
print("使用来源：", a_bundle["origin"])
print("内容指纹：", a_bundle["content_fingerprint"])
````

## 2. 检查文件能否关联，自动列出还没说明的部分

规则已经有 RULE-001 等编号，不需要 A 重新编。缺的是每条规则里面“具体要核实哪件事”的编号和说明。

现有义务文件的一个来源要求块可能包含几件事，所以程序只显示“第几个块”用于定位，绝不把数组序号当正式义务编号。支持先取得材料再核查，但不根据数据库名称猜测业务事实。

````python
def inspect_a_bundle(bundle):
    rules, groups = bundle["rules"], bundle["obligations"]
    if not isinstance(rules, list) or not isinstance(groups, list):
        raise ValueError("两份A组文件的最外层都必须是列表")
    rule_map, group_map, pending = {}, {}, []
    for rule in rules:
        if not isinstance(rule, dict) or not isinstance(rule.get("rule_id"), str):
            raise ValueError("每条规则必须有文字 rule_id")
        rid = rule["rule_id"]
        if not rid or rid in rule_map:
            raise ValueError("规则编号为空或重复：" + rid)
        for key in ["rule_name", "priority", "scope", "effective_time", "thresholds", "exceptions"]:
            if key not in rule:
                raise ValueError(rid + " 缺少字段 " + key)
        if type(rule["priority"]) is not int:
            raise ValueError(rid + " 的 priority 必须是整数")
        if not isinstance(rule["exceptions"], list) or not all(isinstance(x, str) for x in rule["exceptions"]):
            raise ValueError(rid + " 的 exceptions 必须是文字列表")
        for key in ["customer_tier", "plan_type"]:
            v = rule["scope"].get(key)
            if not isinstance(v, list) or not v or not all(isinstance(x,str) and x for x in v):
                raise ValueError(rid + " 的范围字段有误：" + key)
        period = rule["effective_time"]
        # 当前实际JSON使用start/end；不会静默混用字段表中的另一个名字。
        if not isinstance(period, dict) or not all(k in period for k in ["start", "end"]):
            raise ValueError(rid + " 有效期需要 start/end；请与字段表统一")
        if parse_checker_time(period["start"]) > parse_checker_time(period["end"]):
            raise ValueError(rid + " 生效时间晚于失效时间")
        if not valid_number(rule["thresholds"].get("min_incident_minutes"), positive=True):
            raise ValueError(rid + " 时长门槛不合法")
        rule_map[rid] = rule
    for group in groups:
        if not isinstance(group, dict):
            raise ValueError("义务分组必须是字典")
        rid = group.get("rule_id")
        if rid not in rule_map or rid in group_map:
            raise ValueError("义务对应的规则不存在或重复：" + str(rid))
        blocks = group.get("proof_obligations")
        if not isinstance(blocks, list) or not blocks:
            raise ValueError(str(rid) + " 缺少义务列表")
        group_map[rid] = blocks
        for index, block in enumerate(blocks, 1):
            if not isinstance(block, dict):
                raise ValueError("义务内容必须是字典")
            sources = block.get("required_sources")
            if not isinstance(sources, list) or not sources or not all(isinstance(x,str) and x for x in sources):
                raise ValueError(rid + " 缺少明确的数据来源")
            deps = block.get("dependencies", {})
            if not isinstance(deps, dict):
                raise ValueError("dependencies必须是字典")
            missing = [k for k in ["obligation_id", "statement", "required_for_decision", "must_verify"] if k not in block]
            if missing:
                pending.append({"rule_id": rid, "block_position_for_review_only": index,
                                "sources": sources, "missing_fields": missing,
                                "message": "请A明确这里要核实几件事，必要时拆成多项。"})
    for rid in rule_map:
        if rid not in group_map:
            pending.append({"rule_id":rid,"message":"缺少对应证据要求"})
    return rule_map, group_map, pending

a_rule_map, a_group_map, a_pending = inspect_a_bundle(a_bundle)
for rid, rule in a_rule_map.items():
    print(rid, rule["rule_name"], "门槛：", rule["thresholds"]["min_incident_minutes"],
          "现有来源要求块：", len(a_group_map.get(rid, [])))
print("需要补充说明的来源要求块：", len(a_pending))
````

## 3. 用实际规则生成检查清单和数值预检查

这一格读取A的真实门槛，不再统一写死60分钟。只有明确选定 rule_id 后才做预检查，不自动猜客户应选哪条规则。

**“原始时长达到门槛”不等于“有效时长达标”，更不等于“可以赔偿”。** 例如90分钟可能还要扣维护。这里输出数值比较与待确认项，不把它包装成最终SUPPORT。

试用规则的999999和ALL_EXCEPTIONS具有特殊业务意义，暂不把它当普通门槛执行。

````python
def review_with_a_rule(bundle, rule_id, task, records):
    rule_map, group_map, pending = inspect_a_bundle(bundle)
    if rule_id not in rule_map:
        raise ValueError("不存在这个规则编号：" + str(rule_id))
    rule = rule_map[rule_id]
    blocks = group_map.get(rule_id, [])
    # 由已有来源声明形成清单，保留原来的依赖说明供Loop和A一起核对。
    requirements = [{"review_position": i, "required_sources": deepcopy(block["required_sources"]),
                     "dependencies": deepcopy(block.get("dependencies", {})),
                     "business_status": "UNKNOWN",
                     "reason": "来源要求已读取；具体命题和成立条件仍需确认。"}
                    for i, block in enumerate(blocks, 1)]
    config = deepcopy(demo_config)
    config.update(rule_id=rule_id, threshold=rule["thresholds"]["min_incident_minutes"],
                  rules_version="content-" + bundle["content_fingerprint"])
    checked = run_duration_checker(task, records, config)
    special = "ALL_EXCEPTIONS" in rule["exceptions"]
    numeric = []
    for audit in checked["audits"]:
        if audit["usable"]:
            record = audit["record"]
            numeric.append({"evidence_id": record["evidence_id"],
                            "raw_minutes": record["incident_minutes"],
                            "threshold_from_a": config["threshold"],
                            "raw_threshold_met": None if special else record["incident_minutes"] >= config["threshold"],
                            "note": "特殊免赔规则待明确" if special else "仅比较原始时长，尚未计算维护扣除"})
    return {"task_id": task["task_id"], "rule_id": rule_id, "synthetic_demo": True,
            "a_file_origin": bundle["origin"], "a_content_fingerprint": bundle["content_fingerprint"],
            "selection_status": "CALLER_SELECTED_NOT_BUSINESS_VALIDATED",
            "scope_from_a": deepcopy(rule["scope"]),
            "effective_time_from_a": deepcopy(rule["effective_time"]),
            "priority_from_a": rule["priority"], "exceptions_from_a": deepcopy(rule["exceptions"]),
            "requirements_from_a": requirements, "numeric_prechecks": numeric,
            "evidence_audits": checked["audits"],
            "definition_gaps": [p for p in pending if p["rule_id"] == rule_id],
            "current_decision": None,
            "blocking_reasons": ["RULE_APPLICABILITY_PENDING", "OBLIGATION_SEMANTICS_PENDING",
                                 "MAINTENANCE_AND_CONTRACT_POLICY_PENDING"]}

# 显式指定普通客户规则；这只是集成演示，不声称C001已证明属于普通客户。
a_review = review_with_a_rule(a_bundle, "RULE-002", demo_task, [demo_record])
print(json.dumps(a_review, ensure_ascii=False, indent=2))
````

## 4. 验证文件接入真的生效
验证不仅“读到了文件”，还要看改变文件里的门槛后，程序是否跟着改变；并确认坏格式、未知规则、特殊规则不会被默默通过。

````python
assert len(a_rule_map) == 5  # 当前内嵌版本有5条规则；后续新增规则时请同步修改本例断言。
assert a_review["numeric_prechecks"][0]["threshold_from_a"] == 60
changed_bundle = deepcopy(a_bundle)
next(r for r in changed_bundle["rules"] if r["rule_id"] == "RULE-002")["thresholds"]["min_incident_minutes"] = 100
assert review_with_a_rule(changed_bundle, "RULE-002", demo_task, [demo_record])["numeric_prechecks"][0]["raw_threshold_met"] is False
assert review_with_a_rule(a_bundle, "RULE-005", demo_task, [demo_record])["numeric_prechecks"][0]["raw_threshold_met"] is None
for rid in a_rule_map:
    assert review_with_a_rule(a_bundle, rid, demo_task, [demo_record])["current_decision"] is None
bad = deepcopy(a_bundle)
bad["rules"].append(deepcopy(bad["rules"][0]))
try:
    inspect_a_bundle(bad)
except ValueError:
    pass
else:
    raise AssertionError("重复规则编号没有被拒绝")
assert review_with_a_rule(a_bundle, "RULE-002", demo_task, [changed_record(tool_status="ACCESS_DENIED")])["numeric_prechecks"] == []
print("A文件接入验证通过：门槛读取、改值生效、特殊规则、重复编号、403和整体结论保护。")
````

## 1. 这次把哪些回答变成了程序？

| A的回复 | 程序的做法 |
|---|---|
| 先明确合同，再进行后面的判断 | 合同超时、403、分页未完成时，后续检查保留未执行 |
| 故障当时有效的合同和版本才适用 | 用故障时点过滤有效期；新版本尚未生效时仍使用旧版 |
| 专属合同优先，不能随便补通用条款 | 合同明确不允许即反驳；缺少条款返回未知 |
| VIP确认无适用合同时可走通用政策 | 只有正确范围完整查询的回执才允许回退 |
| 维护只扣重合部分 | 先求交集再合并重叠维护，避免扣两遍 |
| 刚好达到门槛也算满足 | 直接按秒比较，不把59分59秒四舍五入成60分钟 |
| 工单与后台日志核实故障原因 | 两者一致且没有命中免责原因才通过；矛盾时冲突 |
| 有充分反驳可提前结束 | 记录否决依据，其余检查标注因已否决而跳过 |

**文字回复还没有提供真实工具字段。** 下面的输入是我们定义的可替换演示格式，不是声称Loop组已经实现。B-CONTRACT等编号是B侧提议，后续可按团队统一编号映射。

### 暂时采用的工程约定（要在联调时核对）
- 合同生效起止时刻都包含；故障跨合同或规则边界时保守返回未知，不自创分段补偿算法。
- 同一时点出现多份有效专属合同/重叠版本时返回冲突；未编造同级合同的优先顺序。
- PLAN_DB是统一套餐资料的演示来源；金融、KA、试用另查A旧义务清单指定的专门来源。
- 工单记录承载排查报告的结构化根因，BACKEND_LOGS承载后台日志；真实名称及原因映射待Loop B提供。
- 维护是否扣除必须在专属条款里明确；无专属合同时，RULE-002按现有规则未规定维护扣减处理。如A更新规则，应同步改动。
- 完整查询里的无效材料可以排除，但记录元数据缺失、查询范围不正确或分页未完，不能证明没有适用合同。
- 此版只处理同一客户、同一服务、同一次故障，不处理金额、多次故障累计和不明确的特殊身份回退。


````python
# 第四阶段可以独立运行：读取本地规则或使用你已提供的规则副本。
import json
from pathlib import Path
V2_A_RULES_COPY = json.loads(r'''[
  {
    "rule_id": "RULE-001",
    "rule_name": "VIP客户专属SLA补偿规则",
    "description": "针对购买企业VIP套餐的客户，故障满30分钟即可申请赔偿，但需扣除计划内维护；若因客户配置错误或不可抗力导致的故障属于免责例外。",
    "priority": 1,
    "effective_time": {
      "start": "2026-01-01T00:00:00Z",
      "end": "2026-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": ["VIP"],
      "plan_type": ["Enterprise_Pro"]
    },
    "thresholds": {
      "min_incident_minutes": 30
    },
    "exceptions": [
      "CLIENT_CONFIG_ERROR",
      "FORCE_MAJEURE"
    ]
  },
  {
    "rule_id": "RULE-002",
    "rule_name": "通用客户标准补偿政策",
    "description": "针对普通基础套餐客户，无专属合同下降级适用；故障满60分钟可赔偿；同样排除客户自身原因引发的故障。",
    "priority": 100,
    "effective_time": {
      "start": "2025-01-01T00:00:00Z",
      "end": "2027-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": ["STANDARD", "BASIC"],
      "plan_type": ["ALL"]
    },
    "thresholds": {
      "min_incident_minutes": 60
    },
    "exceptions": [
      "CLIENT_CONFIG_ERROR"
    ]
  },
  {
    "rule_id": "RULE-003",
    "rule_name": "金融专区高可用保障规则",
    "description": "针对金融专区套餐客户，故障达到15分钟即可赔偿；排除黑客攻击与客户自身配置错误引发的故障。",
    "priority": 5,
    "effective_time": {
      "start": "2026-01-01T00:00:00Z",
      "end": "2026-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": ["FINANCE"],
      "plan_type": ["FINANCE_ZONE"]
    },
    "thresholds": {
      "min_incident_minutes": 15
    },
    "exceptions": [
      "DDoS_ATTACK",
      "CLIENT_CONFIG_ERROR"
    ]
  },
  {
    "rule_id": "RULE-004",
    "rule_name": "战略合作伙伴大客户极速补偿规则",
    "description": "针对战略合作大客户（KA），故障达10分钟即触发补偿保障，优先级别极高；仅排除客户人为主动破坏情况。",
    "priority": 2,
    "effective_time": {
      "start": "2026-01-01T00:00:00Z",
      "end": "2028-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": ["KA_STRATEGIC"],
      "plan_type": ["Custom_Dedicated"]
    },
    "thresholds": {
      "min_incident_minutes": 10
    },
    "exceptions": [
      "CLIENT_MALICIOUS_ACTION"
    ]
  },
  {
    "rule_id": "RULE-005",
    "rule_name": "免费试用与沙盒环境免赔规则",
    "description": "针对使用免费试用版或沙盒测试环境的客户，不提供SLA赔偿承诺，故障门槛设为不适用（极高值），所有故障均属于免赔范围。",
    "priority": 10,
    "effective_time": {
      "start": "2025-01-01T00:00:00Z",
      "end": "2099-12-31T23:59:59Z"
    },
    "scope": {
      "customer_tier": ["TRIAL", "SANDBOX"],
      "plan_type": ["Free_Tier"]
    },
    "thresholds": {
      "min_incident_minutes": 999999
    },
    "exceptions": [
      "ALL_EXCEPTIONS"
    ]
  }
]''')
v2_rules_path = Path("rules.json")
v2_rules = json.loads(v2_rules_path.read_text(encoding="utf-8-sig")) if v2_rules_path.exists() else V2_A_RULES_COPY
print("规则来源：", "当前上传的rules.json" if v2_rules_path.exists() else "此前A组文件副本")
````

## 2. 时间和维护计算工具
这里把时间文字变成可以比较的时间。维护区间先与故障取重合部分，再合并互相重叠的维护。
例如维护10:00—10:40和10:20—10:50，合起来是50分钟，不能扣70分钟。

````python
"""依据A组文字回复的可运行Checker。只接收结构化合成材料，不调用外部工具。"""
from copy import deepcopy
from datetime import datetime
from math import isfinite
import hashlib
import json


class InputError(ValueError):
    """输入格式有误，和业务上的UNKNOWN分开。"""


def time_value(value):
    # 时间必须有时区：Z表示UTC，+08:00表示北京时间。
    if not isinstance(value, str):
        raise InputError("时间必须是文字")
    try:
        result = datetime.fromisoformat(value.replace("Z", "+00:00"))
    except ValueError as error:
        raise InputError("时间格式错误：" + value) from error
    if result.tzinfo is None:
        raise InputError("时间缺少时区：" + value)
    return result


def require_text(obj, key):
    if not isinstance(obj.get(key), str) or not obj[key].strip():
        raise InputError("缺少或填错字段：" + key)
    return obj[key]


def finite_number(value):
    return type(value) in (float, int) and isfinite(value)


def merged_overlap_seconds(start, end, intervals):
    """只计算故障与维护的交集；先合并重叠维护，避免重复扣减。"""
    clipped = []
    for left, right in intervals:
        if right <= left:
            raise InputError("维护结束时间必须晚于开始时间")
        left, right = max(left, start), min(right, end)
        if left < right:
            clipped.append((left, right))
    clipped.sort()
    merged = []
    for left, right in clipped:
        if merged and left <= merged[-1][1]:
            merged[-1] = (merged[-1][0], max(right, merged[-1][1]))
        else:
            merged.append((left, right))
    return sum((right-left).total_seconds() for left, right in merged)


````

## 3. 新的主程序：run_checker_v2

函数接收三样东西：任务、各工具完整的最新返回、A组规则。返回最终结论、每项检查结果、缺口、引用和原始记录。

代码比较长，是因为把四个步骤放在一个入口中：**合同 → 身份/规则/条款 → 净时长 → 免责原因**。内部注释标明了每个步骤。现在不用逐行背，先看后面的完整例子。

`current_decision`：True是可以补偿，False是不补偿，None是暂时无法判断。它与每个检查项的四态是两层不同含义。

Checker只建议补证或人工复核，不实际操作工具，也不声称已转人工。补查后请传入完整的新快照重新调用。

````python
def run_checker_v2(task, responses, rules):
    """每次传入完整最新快照；不在函数内重试、查询或实际转人工。"""
    task, responses, rules = deepcopy(task), deepcopy(responses), deepcopy(rules)
    if not isinstance(task, dict) or not isinstance(responses, dict) or not isinstance(rules, list):
        raise InputError("task/responses需要字典，rules需要列表")
    for key in ['task_id', 'customer_id', 'scope', 'incident_id']:
        require_text(task, key)
    start, end = time_value(task.get('incident_start')), time_value(task.get('incident_end'))
    if end <= start:
        raise InputError("故障结束时间必须晚于开始时间")
    rule_map = {}
    for rule in rules:
        if not isinstance(rule, dict):
            raise InputError("每条规则必须是字典")
        rid = require_text(rule, 'rule_id')
        if rid in rule_map:
            raise InputError("规则编号重复")
        rule_map[rid] = rule
    # 所有查询回执和材料都必须有唯一编号，方便追溯。
    seen = set()
    for source, response in responses.items():
        if not isinstance(response, dict) or not isinstance(response.get('records'), list):
            raise InputError(source + ' 需要records列表，失败时也请填空列表')
        for record in [response] + response['records']:
            if not isinstance(record, dict):
                raise InputError('记录必须是字典')
            eid = require_text(record, 'evidence_id')
            require_text(record, 'citation')
            if eid in seen:
                raise InputError('证据编号重复：' + eid)
            seen.add(eid)

    stages = [
        ('B-CONTRACT', '确认故障当时的适用合同', ['CRM_CONTRACT_DB']),
        ('B-IDENTITY', '确认客户身份和套餐', ['PLAN_DB']),
        ('B-RULE', '确定适用规则和条款', []),
        ('B-PERMISSION', '条款允许本次补偿', []),
        ('B-DURATION', '有效故障时长达到门槛', ['MONITORING_SYSTEM']),
        ('B-CAUSE', '故障原因不属于免责范围', ['TICKET_SYSTEM', 'BACKEND_LOGS'])]
    # B前缀编号是实现提议，需团队统一；它们不是A文件原本就有的字段。
    results = {oid: {'obligation_id': oid, 'statement': statement, 'status': 'UNKNOWN',
                     'reason_codes': ['NOT_CHECKED'], 'reason': '尚未检查', 'evidence_ids': [],
                     'required_sources': sources, 'required_for_decision': True}
               for oid, statement, sources in stages}
    audits, provenance = [], []
    selected_rule = None
    metrics = {}

    def set_result(oid, status, code, reason, ids=()):
        results[oid].update(status=status, reason_codes=[code], reason=reason, evidence_ids=list(ids))

    def finish(decision=None):
        # A明确允许必要条件已被有效反驳时提前结束；仍保留未执行项及其原因。
        if decision is False:
            for item in results.values():
                if item['reason_codes'] == ['NOT_CHECKED']:
                    item.update(reason_codes=['SKIPPED_AFTER_DECISIVE_REFUTE'],
                                reason='已有充分证据否定必要条件，本轮不再检查此项',
                                required_for_decision=False)
        gaps = [deepcopy(x) for x in results.values()
                if x['status'] in ('UNKNOWN', 'CONFLICT') and x['required_for_decision']]
        all_codes = {c for x in gaps for c in x['reason_codes']}
        # 这里只给Loop提示，不能声称已经完成转人工。
        handoff = 'HUMAN_REVIEW_RECOMMENDED' if ('ACCESS_DENIED' in all_codes or 'TIMEOUT' in all_codes
                   or any(x['status'] == 'CONFLICT' for x in gaps)) else 'SUPPLEMENT_OR_CONFIRM'
        if not gaps:
            handoff = 'NO_REQUIRED_GAP'
        payload = {'task': task, 'responses': responses, 'rules': rules}
        return {'schema_version': '02B-demo-v2', 'synthetic_demo': True,
                'task_id': task['task_id'], 'current_decision': decision,
                'checker_decision': {True:'RECOMMEND_COMPENSATION', False:'DO_NOT_RECOMMEND', None:'UNDETERMINED'}[decision],
                'selected_rule_id': selected_rule, 'obligation_results': list(results.values()),
                'gaps': gaps, 'loop_hint': handoff, 'duration': metrics,
                'audits': audits, 'used_policy_evidence': provenance,
                'raw_responses': responses,
                'snapshot_fingerprint': hashlib.sha256(json.dumps(payload,sort_keys=True,ensure_ascii=False).encode()).hexdigest()[:16]}

    def query(source):
        """先验证查询本身。查错客户或没有翻完页，不能证明没有合同。"""
        response = responses.get(source)
        if response is None:
            return None, 'NOT_QUERIED', []
        receipt = [response['evidence_id']]
        if response.get('source_type') != source:
            return None, 'WRONG_QUERY_SOURCE', receipt
        if any(response.get(k) != task[k] for k in ['customer_id','scope','incident_id']):
            return None, 'WRONG_QUERY_SCOPE', receipt
        try:
            if time_value(response.get('incident_start')) != start or time_value(response.get('incident_end')) != end:
                return None, 'WRONG_QUERY_TIME', receipt
            fetched = time_value(response.get('fetched_at'))
            if fetched < end:
                return None, 'QUERY_BEFORE_INCIDENT_END', receipt
        except InputError:
            return None, 'INVALID_QUERY_TIME', receipt
        if response.get('tool_status') != 'SUCCESS':
            code = response.get('tool_status')
            return None, code if code in ['TIMEOUT','ACCESS_DENIED'] else 'TOOL_ERROR', receipt
        if response.get('pagination_complete') is not True:
            return None, 'INCOMPLETE_PAGINATION', receipt
        return response['records'], None, receipt

    def applicable(source, temporal=True):
        records, error, receipt = query(source)
        if error:
            return [], error, receipt
        valid, malformed = [], False
        for record in records:
            reasons = []
            if record.get('source_type') != source:
                reasons.append('WRONG_RECORD_SOURCE'); malformed = True
            # 完整且范围正确的查询中，不适用记录可排除；保留审计说明。
            if any(record.get(k) != task[k] for k in ['customer_id','scope']):
                reasons.append('RECORD_NOT_APPLICABLE')
                # 缺失身份字段不是已确认“不属于该客户”。
                if any(not isinstance(record.get(k),str) for k in ['customer_id','scope']):
                    malformed = True
            if not isinstance(record.get('version'), str) or not record['version'].strip():
                reasons.append('MISSING_VERSION'); malformed = True
            if temporal:
                try:
                    left, right = time_value(record.get('effective_from')), time_value(record.get('effective_to'))
                    if left > right:
                        raise InputError('有效期顺序错误')
                    # 演示边界约定：包含起止时刻；跨版本故障保守返回未知。
                    if not left <= start <= right:
                        reasons.append('OUTSIDE_EFFECTIVE_TIME')
                    elif end > right:
                        reasons.append('INCIDENT_CROSSES_EFFECTIVE_BOUNDARY'); malformed = True
                except InputError:
                    reasons.append('INVALID_EFFECTIVE_TIME'); malformed = True
            audits.append({'evidence_id':record['evidence_id'], 'source_type':source,
                           'usable':not reasons, 'reason_codes':reasons})
            if not reasons:
                valid.append(record)
        # 合同中只要有无法判定有效性的记录，就不能以另一个版本或空结果替代。
        return valid, 'UNRESOLVED_RECORD_METADATA' if malformed else None, receipt

    # 第一关：合同必须查询完整。未知时逻辑上阻断后续检查。
    contracts, error, receipt = applicable('CRM_CONTRACT_DB')
    if error:
        set_result('B-CONTRACT','UNKNOWN',error,'合同查询尚未查清，不能使用通用政策兜底',receipt)
        return finish()
    if len(contracts) > 1:
        set_result('B-CONTRACT','CONFLICT','MULTIPLE_EFFECTIVE_CONTRACTS',
                   '故障时点有多份适用合同或重叠版本，尚无可执行的同级裁决规则', [x['evidence_id'] for x in contracts])
        return finish()
    contract = contracts[0] if contracts else None
    contract_ids = [contract['evidence_id']] if contract else receipt
    set_result('B-CONTRACT','SUPPORT','CONTRACT_RESOLVED',
               '已取得唯一适用合同' if contract else '正确范围内完整查询，确认没有适用合同',contract_ids)

    # 第二关：身份资料也必须是故障当时适用的版本。
    identities, error, ids = applicable('PLAN_DB')
    if error or len(identities) != 1:
        state = 'CONFLICT' if len(identities)>1 and not error else 'UNKNOWN'
        set_result('B-IDENTITY',state,error or 'IDENTITY_NOT_UNIQUE','客户身份和套餐尚未明确',ids+[x['evidence_id'] for x in identities])
        return finish()
    identity = identities[0]
    tier, plan = identity.get('customer_tier'), identity.get('plan_type')
    if not isinstance(tier,str) or not isinstance(plan,str):
        set_result('B-IDENTITY','UNKNOWN','MISSING_IDENTITY_FIELDS','缺少客户等级或套餐',[identity['evidence_id']])
        return finish()
    set_result('B-IDENTITY','SUPPORT','IDENTITY_RESOLVED','客户等级和套餐已取得',[identity['evidence_id']])

    # 合同明确指定规则；没有合同时只对A明确的普通/VIP通用回退执行。
    if contract:
        rid = contract.get('rule_id')
    elif tier in ('VIP','STANDARD','BASIC'):
        rid = 'RULE-002'
    elif tier in ('TRIAL','SANDBOX'):
        rid = 'RULE-005'
    else:
        set_result('B-RULE','UNKNOWN','FALLBACK_NOT_DEFINED','该身份无合同时的回退路线尚未明确',receipt)
        return finish()
    if rid not in rule_map:
        set_result('B-RULE','UNKNOWN','RULE_NOT_FOUND','合同关联规则不存在，不能猜测适用规则',contract_ids)
        return finish()
    rule = rule_map[rid]
    try:
        left=time_value(rule['effective_time']['start']);right=time_value(rule['effective_time']['end'])
        tiers=rule['scope']['customer_tier'];plans=rule['scope']['plan_type']
        match = tier in tiers and (plan in plans or 'ALL' in plans)
        if not contract and tier=='VIP' and rid=='RULE-002':
            match = True  # A明确：VIP在确认无适用合同后可按普通客户政策处理。
        if not match or not left <= start <= end <= right:
            set_result('B-RULE','UNKNOWN','RULE_SCOPE_OR_TIME_MISMATCH','尚未锁定适用于本客户和故障时间的规则',contract_ids)
            return finish()
    except (KeyError,TypeError,InputError):
        set_result('B-RULE','UNKNOWN','INVALID_RULE_FIELDS','规则范围或时间字段不完整')
        return finish()
    # 金融/KA/试用身份必须额外取得原义务文件要求的专门来源。
    special_source={'RULE-003':'FINANCE_ZONE_DB','RULE-004':'KA_CLIENT_DB','RULE-005':'TRIAL_ACCOUNTS_DB'}.get(rid)
    special_ids=[]
    if special_source:
        special_records,error,special_ids=applicable(special_source)
        if error or len(special_records)!=1 or special_records[0].get('eligible') is not True:
            set_result('B-RULE','UNKNOWN',error or 'SPECIAL_IDENTITY_UNPROVEN','专门身份资料尚未充分确认',special_ids)
            return finish()
        special_ids += [special_records[0]['evidence_id']]
    selected_rule=rid
    policy_id='POLICY:'+rid
    provenance.append({'evidence_id':policy_id,'citation':'A组 rules.json#'+rid,'rule':deepcopy(rule)})
    set_result('B-RULE','SUPPORT','RULE_LOCKED','已锁定规则；专属合同条款优先',contract_ids+special_ids+[policy_id])

    if rid=='RULE-005':
        # 不拿999999当正常阈值，而按A原规则文字“不提供SLA赔偿”执行。
        set_result('B-PERMISSION','REFUTE','TRIAL_NO_COMPENSATION','适用免费试用/沙盒免赔规则',special_ids+[policy_id])
        return finish(False)
    if contract:
        terms=contract.get('terms')
        if not isinstance(terms,dict) or type(terms.get('allows_compensation')) is not bool:
            set_result('B-PERMISSION','UNKNOWN','MISSING_CONTRACT_TERMS','合同没有明确补偿许可，不能用通用条款填空',contract_ids)
            return finish()
        if not terms['allows_compensation']:
            set_result('B-PERMISSION','REFUTE','EXCLUSIVE_CONTRACT_DENIES','有效专属合同明确不允许本次补偿',contract_ids)
            return finish(False)
        # 必须明确条款，不擅自继承通用政策缺省值。
        threshold=terms.get('min_incident_minutes');deduct=terms.get('deduct_maintenance');exceptions=terms.get('exceptions')
    else:
        threshold=rule.get('thresholds',{}).get('min_incident_minutes')
        exceptions=rule.get('exceptions')
        # RULE-002描述未要求维护扣除；A回复要求按每条规则执行。
        deduct=False
    if not finite_number(threshold) or threshold<=0 or type(deduct) is not bool or not isinstance(exceptions,list) or not all(isinstance(x,str) and x for x in exceptions):
        set_result('B-PERMISSION','UNKNOWN','INCOMPLETE_COMPENSATION_TERMS','门槛、维护处理或免责清单未明确，禁止静默填补',contract_ids+[policy_id])
        return finish()
    if 'ALL_EXCEPTIONS' in exceptions:
        set_result('B-PERMISSION','UNKNOWN','UNSUPPORTED_EXCEPTION_WILDCARD','通配免责条款需要明确含义，不能当普通原因代码处理',contract_ids)
        return finish()
    set_result('B-PERMISSION','SUPPORT','COMPENSATION_TERMS_PRESENT','已取得明确补偿条件，仍需核实故障事实',contract_ids+[policy_id])

    # 第三关：故障区间必须与任务一致，不能拿其他事故的时长。
    incidents,error,ids=applicable('MONITORING_SYSTEM',temporal=False)
    if error or not incidents:
        set_result('B-DURATION','UNKNOWN',error or 'MISSING_INCIDENT','缺少有效故障记录',ids)
        return finish()
    for record in incidents:
        try:
            if record.get('incident_id') != task['incident_id'] or time_value(record.get('start'))!=start or time_value(record.get('end'))!=end:
                raise InputError('故障记录与任务区间不一致')
        except InputError:
            set_result('B-DURATION','CONFLICT','INCIDENT_TIME_CONFLICT','故障区间与任务不一致，需核实',[x['evidence_id'] for x in incidents])
            return finish()
    evidence_ids=[x['evidence_id'] for x in incidents]
    maintenance_seconds=0
    if deduct:
        results['B-DURATION']['required_sources'].append('MAINTENANCE_CALENDAR')
        maintenance,error,receipts=applicable('MAINTENANCE_CALENDAR',temporal=False)
        if error:
            set_result('B-DURATION','UNKNOWN',error,'维护查询未完成，不能确认净故障时长',receipts)
            return finish()
        evidence_ids+=receipts
        intervals=[]
        for record in maintenance:
            try:
                if type(record.get('planned')) is not bool:
                    raise InputError('缺少是否计划维护的标记')
                left,right=time_value(record.get('start')),time_value(record.get('end'))
                if right<=left:
                    raise InputError('维护区间错误')
                if record['planned']:
                    intervals.append((left,right))
                evidence_ids.append(record['evidence_id'])
            except InputError:
                set_result('B-DURATION','UNKNOWN','INVALID_MAINTENANCE_RECORD','维护时间或属性不明确，不能计算净时长',receipts)
                return finish()
        maintenance_seconds=merged_overlap_seconds(start,end,intervals)
    total_seconds=(end-start).total_seconds()
    net_seconds=total_seconds-maintenance_seconds
    metrics.update(raw_minutes=total_seconds/60,maintenance_minutes=maintenance_seconds/60,
                   net_minutes=net_seconds/60,threshold_minutes=threshold)
    met=net_seconds >= threshold*60
    set_result('B-DURATION','SUPPORT' if met else 'REFUTE',
               'DURATION_THRESHOLD_MET' if met else 'DURATION_BELOW_THRESHOLD',
               f'总时长{total_seconds/60:g}分钟，扣维护{maintenance_seconds/60:g}分钟，净时长{net_seconds/60:g}分钟，门槛{threshold:g}分钟',evidence_ids)
    if not met:
        return finish(False)

    # 第四关：工单/排查报告与后台日志需共同核实根因。
    # 工单记录承载排查报告结果，真实字段仍需要Loop B给适配样例。
    cause_values=[];cause_ids=[]
    for source in ['TICKET_SYSTEM','BACKEND_LOGS']:
        records,error,ids=applicable(source,temporal=False)
        if error or not records:
            set_result('B-CAUSE','UNKNOWN',error or 'MISSING_CAUSE_EVIDENCE','故障原因资料不齐，不能默认没有免责原因',ids)
            return finish()
        for record in records:
            causes=record.get('cause_codes')
            if record.get('incident_id')!=task['incident_id'] or not isinstance(causes,list) or not causes or not all(isinstance(x,str) and x for x in causes):
                set_result('B-CAUSE','UNKNOWN','INVALID_CAUSE_RECORD','需要同一故障明确的非空根因列表',[record['evidence_id']])
                return finish()
            # 演示适配层使用明确的原因枚举；陌生文字不能自动算作非免责。
            known_causes={'PROVIDER_FAILURE','CLIENT_CONFIG_ERROR','FORCE_MAJEURE','DDoS_ATTACK','CLIENT_MALICIOUS_ACTION'}
            if any(c not in known_causes for c in causes):
                set_result('B-CAUSE','UNKNOWN','UNMAPPED_CAUSE_CODE','原因代码尚未映射，请工具组确认含义',[record['evidence_id']])
                return finish()
            cause_values.append(set(causes));cause_ids.append(record['evidence_id'])
    if any(value!=cause_values[0] for value in cause_values):
        set_result('B-CAUSE','CONFLICT','CAUSE_CONFLICT','工单与后台日志对根因的描述不一致，不能自行挑一份',cause_ids)
        return finish()
    excluded=bool(cause_values[0]&set(exceptions))
    set_result('B-CAUSE','REFUTE' if excluded else 'SUPPORT',
               'EXCLUDED_CAUSE' if excluded else 'NO_EXCLUDED_CAUSE',
               '已命中适用免责原因' if excluded else '两类资料一致，根因未命中免责清单',cause_ids)
    return finish(not excluded)
````

## 4. 准备模拟数据和边界测试
样例客户C001是VIP，故障90分钟、计划维护40分钟，合同门槛30分钟，根因是提供商故障。完整情况下净时长50分钟，可得到“可以补偿”。

每次工具返回包含一个查询回执和records列表。即使查询失败也返回空列表；查询成功且全部翻页完成的空列表，才能作为“没有适用记录”的依据。

下面先定义样例和测试，暂不执行测试。它们是开发自测，不是测试组的独立标准答案。

````python
"""合成开发自测：不是测试组独立真值，也不是模型准确率。"""
from copy import deepcopy
import json
from pathlib import Path


def make_fixture():
    task={'task_id':'DEMO-TASK-V2','customer_id':'C001','scope':'CLOUD_SERVER',
          'incident_id':'INC-001','incident_start':'2026-06-01T10:00:00Z','incident_end':'2026-06-01T11:30:00Z'}
    def record(source,eid,**kwargs):
        return dict(evidence_id=eid,citation='demo://'+eid,source_type=source,
                    customer_id='C001',scope='CLOUD_SERVER',version='V1',
                    effective_from='2026-01-01T00:00:00Z',effective_to='2026-12-31T23:59:59Z',**kwargs)
    contract=record('CRM_CONTRACT_DB','E-CONTRACT',rule_id='RULE-001',
                    terms={'allows_compensation':True,'min_incident_minutes':30,
                           'deduct_maintenance':True,'exceptions':['CLIENT_CONFIG_ERROR','FORCE_MAJEURE']})
    material={
        'CRM_CONTRACT_DB':[contract],
        'PLAN_DB':[record('PLAN_DB','E-PLAN',customer_tier='VIP',plan_type='Enterprise_Pro')],
        'MONITORING_SYSTEM':[record('MONITORING_SYSTEM','E-INCIDENT',incident_id='INC-001',start=task['incident_start'],end=task['incident_end'])],
        'MAINTENANCE_CALENDAR':[record('MAINTENANCE_CALENDAR','E-MAINT-1',planned=True,start='2026-06-01T10:00:00Z',end='2026-06-01T10:40:00Z')],
        'TICKET_SYSTEM':[record('TICKET_SYSTEM','E-TICKET',incident_id='INC-001',cause_codes=['PROVIDER_FAILURE'])],
        'BACKEND_LOGS':[record('BACKEND_LOGS','E-LOG',incident_id='INC-001',cause_codes=['PROVIDER_FAILURE'])]}
    responses={}
    for source,records in material.items():
        responses[source]=dict(task, evidence_id='Q-'+source,citation='demo://query/'+source,
                               source_type=source,tool_status='SUCCESS',pagination_complete=True,
                               fetched_at='2026-06-02T00:00:00Z',records=records)
    return task,responses


def run_checks(rules):
    """给每种边界单独保存结果，方便开会解释。"""
    table=[]
    def check(name,change,expected,extra=None):
        task,responses=make_fixture()
        change(task,responses)
        before=deepcopy((task,responses,rules))
        result=run_checker_v2(task,responses,rules)
        assert result['current_decision'] is expected,(name,result)
        assert before==(task,responses,rules),'函数修改了输入'
        assert result==run_checker_v2(task,responses,rules),'结果不确定'
        if extra:extra(result)
        table.append({'case':name,'expected':expected,'actual':result['current_decision'],'passed':True})
    def rec(r,source):return r[source]['records'][0]
    def setv(source,key,value):return lambda t,r:r[source].update({key:value})
    def contract_terms(key,value):return lambda t,r:rec(r,'CRM_CONTRACT_DB')['terms'].update({key:value})
    check('VIP正常：90扣40得到50',lambda t,r:None,True,lambda x:assert_net(x,50))
    check('刚好30分钟',lambda t,r:rec(r,'MAINTENANCE_CALENDAR').update(end='2026-06-01T11:00:00Z'),True,lambda x:assert_net(x,30))
    check('净时长20分钟',lambda t,r:rec(r,'MAINTENANCE_CALENDAR').update(end='2026-06-01T11:10:00Z'),False)
    check('差一秒不达标',lambda t,r:rec(r,'MAINTENANCE_CALENDAR').update(end='2026-06-01T11:00:01Z'),False)
    check('重叠维护不重复扣',lambda t,r:r['MAINTENANCE_CALENDAR']['records'].append(dict(rec(r,'MAINTENANCE_CALENDAR'),evidence_id='E-MAINT-2',start='2026-06-01T10:20:00Z',end='2026-06-01T10:50:00Z')),True,lambda x:assert_net(x,40))
    check('维护跨出故障区间只扣交集',lambda t,r:rec(r,'MAINTENANCE_CALENDAR').update(start='2026-06-01T09:00:00Z',end='2026-06-01T10:40:00Z'),True,lambda x:assert_net(x,50))
    check('完整维护空结果可不扣',setv('MAINTENANCE_CALENDAR','records',[]),True,lambda x:assert_net(x,90))
    check('非计划维护不扣',lambda t,r:rec(r,'MAINTENANCE_CALENDAR').update(planned=False),True,lambda x:assert_net(x,90))
    check('维护分页未完',setv('MAINTENANCE_CALENDAR','pagination_complete',False),None)
    check('维护时间反向',lambda t,r:rec(r,'MAINTENANCE_CALENDAR').update(end='2026-06-01T09:00:00Z'),None)
    check('维护缺时间',lambda t,r:rec(r,'MAINTENANCE_CALENDAR').pop('start'),None)
    check('合同未查询',lambda t,r:r.pop('CRM_CONTRACT_DB'),None)
    check('合同超时',setv('CRM_CONTRACT_DB','tool_status','TIMEOUT'),None)
    check('合同403',setv('CRM_CONTRACT_DB','tool_status','ACCESS_DENIED'),None)
    check('合同查询错客户',setv('CRM_CONTRACT_DB','customer_id','C002'),None)
    check('合同空第一页不回退',lambda t,r:r['CRM_CONTRACT_DB'].update(records=[],pagination_complete=False),None)
    check('合同查询时间错',setv('CRM_CONTRACT_DB','incident_start','2026-06-02T10:00:00Z'),None)
    check('合同回执缺抓取时间',lambda t,r:r['CRM_CONTRACT_DB'].pop('fetched_at'),None)
    check('完整合同空结果：VIP回退',setv('CRM_CONTRACT_DB','records',[]),True)
    check('完整查询仅过期合同：回退',lambda t,r:rec(r,'CRM_CONTRACT_DB').update(effective_to='2026-05-01T00:00:00Z'),True)
    check('完整查询仅不适用客户记录',lambda t,r:rec(r,'CRM_CONTRACT_DB').update(customer_id='C002'),True)
    check('合同缺有效期不能当不存在',lambda t,r:rec(r,'CRM_CONTRACT_DB').pop('effective_to'),None)
    check('合同缺客户不能当不存在',lambda t,r:rec(r,'CRM_CONTRACT_DB').pop('customer_id'),None)
    check('合同明确禁止优先于通用',contract_terms('allows_compensation',False),False)
    check('合同禁止无需后续资料',lambda t,r:(rec(r,'CRM_CONTRACT_DB')['terms'].update(allows_compensation=False),r.pop('MONITORING_SYSTEM'),r.pop('TICKET_SYSTEM')),False)
    check('合同没有维护约定不自动填',lambda t,r:rec(r,'CRM_CONTRACT_DB')['terms'].pop('deduct_maintenance'),None)
    check('合同缺免责清单不自动填',lambda t,r:rec(r,'CRM_CONTRACT_DB')['terms'].pop('exceptions'),None)
    check('布尔值门槛无效',contract_terms('min_incident_minutes',True),None)
    check('文字门槛无效',contract_terms('min_incident_minutes','30'),None)
    check('未知规则',lambda t,r:rec(r,'CRM_CONTRACT_DB').update(rule_id='NOT-A-RULE'),None)
    check('合同与身份范围不符',lambda t,r:rec(r,'PLAN_DB').update(customer_tier='STANDARD'),None)
    check('两份同时有效合同',lambda t,r:r['CRM_CONTRACT_DB']['records'].append(dict(rec(r,'CRM_CONTRACT_DB'),evidence_id='E-CONTRACT-2',version='V2')),None)
    def future(t,r):
        old=rec(r,'CRM_CONTRACT_DB')
        newer=deepcopy(old);newer.update(evidence_id='E-CONTRACT-2',version='V2',effective_from='2026-07-01T00:00:00Z')
        newer['terms']['allows_compensation']=False
        r['CRM_CONTRACT_DB']['records'].append(newer)
    check('新版本故障后生效仍用旧版本',future,True)
    check('故障跨有效期边界',lambda t,r:rec(r,'CRM_CONTRACT_DB').update(effective_to='2026-06-01T11:00:00Z'),None)
    check('查询时已过期但故障时有效',lambda t,r:r['CRM_CONTRACT_DB'].update(fetched_at='2027-02-01T00:00:00Z'),True)
    check('监控缺失',lambda t,r:r.pop('MONITORING_SYSTEM'),None)
    check('故障编号不一致',lambda t,r:rec(r,'MONITORING_SYSTEM').update(incident_id='INC-999'),None)
    check('工单缺失',lambda t,r:r.pop('TICKET_SYSTEM'),None)
    check('日志缺失',lambda t,r:r.pop('BACKEND_LOGS'),None)
    check('免责原因明确',lambda t,r:[rec(r,s).update(cause_codes=['CLIENT_CONFIG_ERROR']) for s in ['TICKET_SYSTEM','BACKEND_LOGS']],False)
    check('工单日志根因冲突',lambda t,r:rec(r,'BACKEND_LOGS').update(cause_codes=['CLIENT_CONFIG_ERROR']),None)
    check('空根因不当作没有免责',lambda t,r:rec(r,'TICKET_SYSTEM').update(cause_codes=[]),None)
    def standard(t,r):
        r['CRM_CONTRACT_DB']['records']=[];rec(r,'PLAN_DB').update(customer_tier='STANDARD',plan_type='Basic')
    check('普通客户无合同通用规则',standard,True)
    def special(t,r,rid,tier,plan,source):
        rec(r,'PLAN_DB').update(customer_tier=tier,plan_type=plan)
        rec(r,'CRM_CONTRACT_DB').update(rule_id=rid)
        base=deepcopy(r['PLAN_DB']);base.update(source_type=source,evidence_id='Q-SPECIAL')
        base['records'][0].update(source_type=source,evidence_id='E-SPECIAL',eligible=True)
        r[source]=base
    check('金融专门身份资料齐全',lambda t,r:special(t,r,'RULE-003','FINANCE','FINANCE_ZONE','FINANCE_ZONE_DB'),True)
    check('KA专门身份资料齐全',lambda t,r:special(t,r,'RULE-004','KA_STRATEGIC','Custom_Dedicated','KA_CLIENT_DB'),True)
    check('试用明确免赔',lambda t,r:special(t,r,'RULE-005','TRIAL','Free_Tier','TRIAL_ACCOUNTS_DB'),False)
    check('金融缺专门身份资料',lambda t,r:(rec(r,'PLAN_DB').update(customer_tier='FINANCE',plan_type='FINANCE_ZONE'),rec(r,'CRM_CONTRACT_DB').update(rule_id='RULE-003')),None)
    check('陌生根因不自动放行',lambda t,r:[rec(r,s).update(cause_codes=['UNKNOWN']) for s in ['TICKET_SYSTEM','BACKEND_LOGS']],None)
    check('通配免责不自动放行',contract_terms('exceptions',['ALL_EXCEPTIONS']),None)
    # 结构错误另报错，不伪装成一次正常业务判断。
    for name,modify in [('重复证据编号',lambda t,r:rec(r,'PLAN_DB').update(evidence_id='E-CONTRACT')),
                        ('任务无时区',lambda t,r:t.update(incident_start='2026-06-01T10:00:00'))]:
        t,r=make_fixture();modify(t,r)
        try:run_checker_v2(t,r,rules)
        except InputError:table.append({'case':name,'expected':'InputError','actual':'InputError','passed':True})
        else:raise AssertionError(name)
    return table


def assert_net(result,expected):
    assert result['duration']['net_minutes']==expected,result['duration']


````

## 5. 跑三个能看懂的完整案例
同一客户、同一故障：材料齐全可以赔；修改专属合同为明确禁止就不赔；合同查询403则未知。这里不修改原始样例，每次重新生成数据。

````python
v2_examples = {}
for name in ["材料齐全", "专属合同禁止", "合同无权限"]:
    task_v2, responses_v2 = make_fixture()
    if name == "专属合同禁止":
        responses_v2["CRM_CONTRACT_DB"]["records"][0]["terms"]["allows_compensation"] = False
    elif name == "合同无权限":
        responses_v2["CRM_CONTRACT_DB"]["tool_status"] = "ACCESS_DENIED"
        responses_v2["CRM_CONTRACT_DB"]["records"] = []
    outcome = run_checker_v2(task_v2, responses_v2, v2_rules)
    v2_examples[name] = outcome
    labels = {True: "可以补偿", False: "不能补偿", None: "暂时无法判断"}
    print(name, "→", labels[outcome["current_decision"]])
    print("净时长计算：", outcome["duration"])
    print("仍需解决：", [(g["obligation_id"], g["reason_codes"]) for g in outcome["gaps"]])
    print()
````

## 6. 一次运行全部新测试
测试包含维护重叠、刚好门槛、差一秒、完整空查询、过期合同、错误客户、超时、403、新旧版本、缺条款、根因矛盾、特殊客户等。

## 7. 导出给同学使用的代码和样例
运行这格，会生成 checker_v2_delivery 文件夹。里面有独立checker_v2.py、开发测试、A规则副本、正常输入、三个完整结果和测试报告。

### 修正了哪些问题
- 单份故障记录与任务不一致是UNKNOWN；不能直接称为两份有效证据的CONFLICT。
- 合同客户编号是空文字时，不能当作“不适用合同”并回退。
- 拒绝重复JSON字段、非有限数字以及撞上规则引用编号的证据。
- 命令行失败时输出INPUT_ERROR，避免下游继续使用上一次成功结论。

### 完整评分怎样理解
“合同明确禁止补偿”可以提前结束，后面时长和原因不必查。这些未查项仍留在评分分母，分数可能不到100，但有效否决仍可形成明确结论。分数不是答案正确概率，界面应优先看结论和阻断项。

## 5. 其他组怎么调用

```python
from checker_api import check_payload

```

评分组使用：

```python
from checker_api import make_score_snapshot

```

每次补查后传**全部最新材料**，不是只交新的一条。`process_state`默认RUNNING；是否完成、是否已转人工由Loop明确提供，不能从Checker建议推断成已经发生。

如果输入错误，程序报错，而不是把数据错误伪装成正常低分。输入格式和已有样例见生成目录里的INTERFACE.md和examples。

| 已接通 | 处理方式 |
|---|---|

### 为什么正常查询仍是未知？
工具README明确说合同只有元数据，没有补偿条款。scope也只有客户编号，没有服务范围。我们不能擅自补成“允许补偿”或“适用于云服务器”。维护还缺是否计划维护，套餐、故障、根因工具尚未提供。

因此，“查询→适配→Checker→评分”可以跑通，但完整业务证据尚不充分。当前8个场景都应该保持UNDETERMINED。这是缺证保护起作用，不是把测试结果改成正确答案。
